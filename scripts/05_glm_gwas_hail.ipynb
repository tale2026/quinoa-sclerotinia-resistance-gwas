```python
from pprint import pprint
import pandas as pd
from collections import Counter
```

```python
snp_counts = mt.aggregate_rows(hl.agg.counter(hl.Struct(ref = mt.alleles[0], alt = mt.alleles[1])))
pprint(snp_counts)
```

```python
counts = Counter(snp_counts)
pprint(counts.most_common())
```

```python
data = []
for line in counts.most_common():
    ref = str(line).split(',')[0].split("=")[1].split("'")[1]
    alt = str(line).split(',')[1].split("'")[1]
    Count = str(line).split(',')[2].split(")")[0]
    data.append([ref, alt, Count])

df_snp_mostcommon = pd.DataFrame(data, columns = ["Reference", "Alternative", "Count"])
df_snp_mostcommon.to_csv("Results/Most_Common_SNP.csv", index = False)


```

```python
mt = hl.variant_qc(mt)
```

```python
mt = mt.filter_rows(
    (mt.variant_qc.dp_stats.mean >= 5) & 
    (mt.variant_qc.call_rate >= 0.95) & 
    (hl.min(mt.variant_qc.AF[0], 1 - mt.variant_qc.AF[0]) >= 0.05)  # MAF >= 5%
)

# Verify the number of remaining variants and samples
n_variants = mt.count_rows()
n_samples = mt.count_cols()

print(f"Remaining variants: {n_variants}, Remaining samples: {n_samples}")
```

```python
eigen, pcs, _ = hl.hwe_normalized_pca(mt.GT)
```

```python
pprint(eigen)
```

```python
PCA_TABLE = pcs.to_pandas()
PCA_TABLE['scores'] = PCA_TABLE['scores'].astype(str)
PCA_TABLE['scores'] = PCA_TABLE['scores'].str.replace(r'[\[\]]', '', regex = True)
expanded_columns = PCA_TABLE['scores'].str.split(',', expand = True)
expanded_columns.columns = [f'PC{i+1}' for i in range(expanded_columns.shape[1])]
df_expanded = pd.concat([PCA_TABLE.drop('scores', axis = 1), expanded_columns], axis = 1)
```

```python
df_expanded.to_csv("Results/PCA_Table.csv", index = False)
```
